```python
# Scan more conversations until we find a Completed survey
found = 0
page = 1

while found < 2 and page <= 10:
    body = PureCloudPlatformClientV2.ConversationQuery()
    body.interval = f"{yesterday}T00:00:00Z/{yesterday}T23:59:59Z"
    body.paging = PureCloudPlatformClientV2.PagingSpec()
    body.paging.page_size   = 100
    body.paging.page_number = page

    result = analytics_api.post_analytics_conversations_details_query(body)

    for conv in (result.conversations or []):
        try:
            surveys = quality_api.get_quality_conversation_surveys(conv.conversation_id)
            for s in (surveys or []):
                if s.status == "Completed" and s.answers:
                    found += 1
                    print(f"✅ COMPLETED SURVEY FOUND")
                    print(f"   Conversation : {conv.conversation_id}")
                    print(f"   Agent        : {s.agent}")
                    print(f"   Completed    : {s.completed_date}")
                    print(f"   Status       : {s.status}")
                    print()
                    print("   ANSWERS:")
                    for group in (s.answers.question_group_scores or []):
                        print(f"   Group: {group.question_group_id}")
                        for q in (group.question_scores or []):
                            print(f"     Q: {q.question_id}  →  Score: {q.score}  Selected: {q.answered}")
                    print()
                    if found >= 2:
                        break
        except:
            pass

    page += 1
    if not result.conversations:
        break

if found == 0:
    print("⚠️  No completed surveys found — try a different date")
    print("    Try: yesterday = '2026-06-20'  (Friday)")
```

```python
target_date = "2026-06-19"

# ── Build survey filter: status = Completed ───────────────────────────
survey_predicate = PureCloudPlatformClientV2.SurveyDetailQueryPredicate()
survey_predicate.type      = "dimension"
survey_predicate.dimension = "surveyStatus"
survey_predicate.value     = "Completed"

survey_filter = PureCloudPlatformClientV2.SurveyDetailQueryFilter()
survey_filter.type       = "and"
survey_filter.predicates = [survey_predicate]

# ── Build conversation query with survey filter ───────────────────────
body = PureCloudPlatformClientV2.ConversationQuery()
body.interval        = f"{target_date}T00:00:00Z/{target_date}T23:59:59Z"
body.survey_filters  = [survey_filter]
body.paging          = PureCloudPlatformClientV2.PagingSpec()
body.paging.page_size   = 10  # small — we only need one good result
body.paging.page_number = 1

try:
    result = analytics_api.post_analytics_conversations_details_query(body)
    print(f"✅ Conversations with completed surveys: {result.total_hits}")
    print()

    # ── For each conversation pull the survey ─────────────────────────
    for conv in (result.conversations or [])[:3]:
        print(f"📞 Conversation: {conv.conversation_id}")
        print(f"   Started    : {conv.conversation_start}")

        surveys = quality_api.get_quality_conversation_surveys(conv.conversation_id)
        for s in (surveys or []):
            if s.status == "Completed" and s.answers:
                print(f"   ✅ Survey   : {s.id}")
                print(f"   Form       : {s.survey_form.name if s.survey_form else 'N/A'}")
                print(f"   Completed  : {s.completed_date}")
                print()
                print("   SCORES:")
                for group in (s.answers.question_group_scores or []):
                    for q in (group.question_scores or []):
                        print(f"     {q.question_id}  →  {q.score}")
                print()
                break  # one survey per conversation is enough

except ApiException as e:
    print(f"❌ {e}")
```
